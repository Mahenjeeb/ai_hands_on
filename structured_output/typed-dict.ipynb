### Strctured Output

#### TypedDict

```python
from typing import TypedDict, Annotated
from langchain_ollama import ChatOllama
from langchain_core.prompts import PromptTemplate

class MyOutPut(TypedDict):
    joke: Annotated[str, "Full joke of the specified topic"]
    one_liner: Annotated[str, "respective joke of one-liner"]

SYSTEM_PROMPT = PromptTemplate.from_template(template = """
You are professional stand-up comedian. Tell me joke about {topic}
"""
)
topic = str(input(">> Q"))
format_prompt = SYSTEM_PROMPT.format(topic = topic)
model = ChatOllama(model="qwen2.5:7b")
st_op = model.with_structured_output(MyOutPut)
resp = st_op.invoke(format_prompt)
resp
```

#### Pydantic Model

```python
from pydantic import BaseModel, Field
from langchain_ollama import ChatOllama
from langchain_core.prompts import PromptTemplate

class MyOutPut(BaseModel):
    joke: str = Field(description= "Full joke of the specified topic")
    one_liner: str = Field(description= "respective joke of one-liner")

SYSTEM_PROMPT = PromptTemplate.from_template(template = """
You are professional stand-up comedian. Tell me joke about {topic}
"""
)
topic = str(input(">> Q"))
format_prompt = SYSTEM_PROMPT.format(topic = topic)
model = ChatOllama(model="qwen2.5:7b")
st_op = model.with_structured_output(schema=MyOutPut)
resp = st_op.invoke(format_prompt)
resp
```
