### Prompt Template

```python
from langchain_core.prompts import PromptTemplate
```

```python
system_prompt_template = """You are a solution architect having 15+ years of experience. 
    You Know all the cloud concepts legacy and the latest one.
    Based on {topic} design a architecture which is scalable and
    mainatainable.
    """
```

##### 1. Using PromptTemplate Class

```python
SYSTEM_PROMPT = PromptTemplate(
    template= system_prompt_template,
    input_variables=['topic']
)
topic = str(input(">>You\n"))
SYSTEM_PROMPT.invoke({"topic": topic})
```

##### 2. Using from_template()

```python
SYSTEM_PROMPT = PromptTemplate.from_template(template= system_prompt_template)
topic = str(input(">>You\n"))
formatted_prompt = SYSTEM_PROMPT.format(topic=topic)
print(formatted_prompt)
```

##### 3. Using a file

```python
with open("prompt-template.txt", mode='w',encoding='utf-8') as file:
    file.write(system_prompt_template)
SYSTEM_PROMPT = PromptTemplate.from_file(
    template_file="prompt-template.txt"
)
topic = str(input(">>You\n"))
formatted_prompt = SYSTEM_PROMPT.format(topic=topic)
print(formatted_prompt)
```
