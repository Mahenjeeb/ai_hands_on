### Chat Prompt Template

#### Messages

```python
from langchain_core.messages import HumanMessage, AIMessage, SystemMessage
from langchain_ollama import ChatOllama

model = ChatOllama(model="qwen2.5:7b")
messages = [SystemMessage(content="You are a Help ful assistant"),
            HumanMessage(content="Hey There !"),]
result = model.invoke(messages)
messages.append(AIMessage(content=result.content))
print(messages)
```

#### ChatPromptTemplate

```python
from langchain_core.prompts import ChatPromptTemplate

msg = ChatPromptTemplate([
    ('system', 'You area helpful assistant. your name is {name}'),
    ('human', '{user_input}')
])
et_msg = msg.invoke({'name': 'bob', 'user_input': 'Tell me a joke in one line'})
res = model.invoke(et_msg.messages)
res
```

#### MessagePlaceHolder

```python
from langchain_core.prompts import MessagesPlaceholder

# MessagesPlaceholder - are used to store conversational history
chat_prompt_template = ChatPromptTemplate([
    ('system', 'You area helpful assistant. your name is {name}'),
    ('human', '{user_input}'),
    MessagesPlaceholder(variable_name="history", optional=True)
])
messages = chat_prompt_template.invoke(
    {'history': [
             ('ai', 'I am doing good'),
             ('human','What is blackhole ?')
     ],
     'name': 'bob',
     'user_input': 'Tell me the name of biggest blackhole?',}
)
res = model.invoke(messages.messages)
res
```
