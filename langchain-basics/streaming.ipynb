### Streaming

`create_agent`

```python
from langchain.agents import create_agent
agent = create_agent(
    model="ollama:qwen2.5:7b"
)
```

`stream_mode = 'values'`

```python
stream = agent.stream({"messages": "What is CSI? And how it's calculated"}, stream_mode="values")
for chunk in stream :
    chunk["messages"][-1].pretty_print()
```

`stream_mode = 'messages'`

```python
stream = agent.stream({"messages": "What is CSI? And how it's calculated"}, stream_mode="messages")
for chunk, metadata in stream:
    print(f"{chunk.content}", end="")
```

#### Tool Streaming

`stream_mode = 'custom'`

```python
from langchain_core.tools import tool
from langgraph.config import get_stream_writer
from langchain.agents import create_agent

@tool("Weather")
def get_weather(location: str) -> str:
    """Fetch weather details on given location"""
    writer = get_stream_writer()
    print("Inside weather details")
    writer(f"fetching weather details for {location}")
    writer(f"{location} details retrived")
    return f"{location} is partly cloudy, 23C"

agent = create_agent(
    model="ollama:qwen2.5:7b",
    tools=[get_weather]
)

for chunk in agent.stream({"messages": "What is the weather in bengaluru"}, stream_mode=["custom","values"]):
    print(f"Response: \n {chunk}")
```
