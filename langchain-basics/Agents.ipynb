### Creating and Invoking Agent

```python
from langchain.agents import create_agent
from langchain.tools import tool
from pydantic import BaseModel

@tool
def search(query: str) -> str:
    """Search Reasults For User Query"""
    return f"searched result for {query}"
class ChatResposne(BaseModel):
    result: str
    timeTaken: str
agent = create_agent(
    model="google_genai:gemini-3.5-flash",
    tools=[search],
    system_prompt="You are a helpful assistant",
    response_format=ChatResposne
)
output = agent.invoke({"messages": [{"role":"user", "content": "What is agent state in langchain"}]})
text = output["messages"][-1].content[0]['text']
print(text)
```

### Streaming

```python
from langchain.messages import HumanMessage, AIMessage
from langchain.agents import create_agent
from langchain.tools import tool
from pydantic import BaseModel

@tool
def search(query: str) -> str:
    """Search Reasults For User Query"""
    return f"searched result for {query}"
class ChatResposne(BaseModel):
    result: str
    timeTaken: str
agent = create_agent(
    model="google_genai:gemini-3.5-flash",
    tools=[search],
    system_prompt="You are a helpful assistant",
    response_format=ChatResposne
)
stream = agent.stream_events({"messages": [{"role":"user", "content": "What is agent state in langchain"}]}, version="v3")
for snapshot in stream.values:
    curr_msg = snapshot["messages"][-1]
    if curr_msg.content:
        if isinstance(curr_msg, HumanMessage):
            print(f"User: {curr_msg.content}")
        elif isinstance(curr_msg, AIMessage):
            print(f"Agent: {curr_msg.content}")
    elif curr_msg.tool_calls:
        print(f"Calling Tools: {[tc['name'] for tc in curr_msg.tool_calls]}")
    
```
