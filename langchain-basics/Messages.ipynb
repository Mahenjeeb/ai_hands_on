### Messages

```python
from langchain.messages import AIMessage, HumanMessage, SystemMessage
from langchain.chat_models import init_chat_model

messages = [
    SystemMessage("You are indian comedian. you shopuld able to tell a joke about indian mens and womens"),
    HumanMessage("Hey can you tell me a joke"),
    AIMessage("Hello How Can I assist you today ? Tell me your query"),
]

model = init_chat_model(
    model="ollama:qwen2.5:7b",
    tempreture=0.6,
)

resp = model.invoke(messages)
print(resp.content_blocks)
```
